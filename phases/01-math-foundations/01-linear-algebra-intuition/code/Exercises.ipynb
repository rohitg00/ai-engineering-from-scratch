1. Implement `Vector.angle_between(other)` that returns the angle in degrees between two vectors

```python
from vectors import Vector

a = Vector([1, 2, 3])
b = Vector([4, 5, 6])

print(f"Angle between = {a.angle_between(b):.4f}")
```

2. Create a 2D scaling matrix that doubles the x-coordinate and triples the y-coordinate, then apply it to the vector [1, 1]

```python
from vectors import Matrix

scalling = Matrix([[2, 0], [0, 3]])
point = Vector([1, 1])

scaled = scalling @ point
print(f"Original: {point}")
print(f"Scalling: {scaled}")
```

3. Given 5 random word-like vectors (dimension 50), find the two most similar using cosine similarity

```python
# Buat 5 vector acak dengan dimensi 50
import random

random.seed(42)
vectors = []
for i in range(5):
    components = [random.uniform(-10, 10) for _ in range(50)]
    v = Vector(components)
    vectors.append(v)
    print(f"Vector {i+1}: dim={v.dim}, magnitude={v.magnitude():.4f}")
```

```python
# Hitung cosine similarity untuk semua pasangan
for i in range(5):
    for j in range(i+1, 5):
        sim = vectors[i].cosine_similarity(vectors[j])
        angle = vectors[i].angle_between(vectors[j])
        print(f"cos(v{i+1}, v{j+1}) = {sim:.6f}  |  angle = {angle:.2f}°")

print("\n" + "="*60)
print("MENCARI PASANGAN DENGAN COSINE SIMILARITY TERTINGGI:")
print("="*60)

# Cari pasangan dengan cosine similarity tertinggi
max_sim = -1
best_pair = None

for i in range(5):
    for j in range(i+1, 5):
        sim = vectors[i].cosine_similarity(vectors[j])
        if sim > max_sim:
            max_sim = sim
            best_pair = (i, j)

i, j = best_pair
print(f"\nPasangan paling mirip: Vector {i+1} dan Vector {j+1}")
print(f"Cosine Similarity: {max_sim:.6f}")
print(f"Angle: {vectors[i].angle_between(vectors[j]):.2f}°")

# Tampilkan detail vector tersebut
print(f"\n--- Detail Vector {i+1} ---")
print(f"Magnitude: {vectors[i].magnitude():.4f}")
print(f"Components (first 10): {vectors[i].components[:10]}")

print(f"\n--- Detail Vector {j+1} ---")
print(f"Magnitude: {vectors[j].magnitude():.4f}")
print(f"Components (first 10): {vectors[j].components[:10]}")
```
